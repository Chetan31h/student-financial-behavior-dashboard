```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("raw_dataset.csv")
```

```python
print(df.head)
print(df.shape)
```

```python
print(df.info())
```

```python
df = df.rename(columns={
    "Timestamp": "Timestamp",
    "Age": "Age",
    "Gender": "Gender",
    "Year of study": "Year",
    "Course/Program": "Course",
    "Branch": "Branch",
    "Do you have bank Account": "Bank_Account",
    "Name of bank where you have account": "Bank_Name",
    "Do you have DEMAT/Trading Account?": "Demat_Account",
    "Name of your broker where you have DEMAT account": "Broker_Name",
    "If you have DEMAT account then in which segment you trade?": "Trading_Segment",
    "What is monthly income ?": "Income",
    "Tell your source of income ": "Income_Source",
    "What is your monthly expenses?": "Expenses",
    "  Do you track your expenses?  ": "Track_Expenses",
    "  Do you understand the concept of budgeting?  ": "Budgeting_Knowledge",
    "What is your average monthly Savings?": "Savings"
})
```

```python


# Check missing values
df.isnull().sum()

# Check unique values (important for Yes/No columns)
for col in df.columns:
    print(col, ":", df[col].unique())
```

```python
df = df.drop(columns=["Timestamp"])
```

```python
df["Branch"] = df["Branch"].fillna("Not Applicable")
```

```python
df["Bank_Name"] = df["Bank_Name"].fillna("Not Available")
```

```python
df["Demat_Account"] = df["Demat_Account"].fillna("No")
```

```python
df["Broker_Name"] = df["Broker_Name"].fillna("Not Applicable")
```

```python
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')
df = df.dropna(subset=['Age'])
df = df[(df['Age'] >= 18) & (df['Age'] <= 26)]
```

```python
df = df[df['Savings'] <= 10000]
```

```python
df.info()
df.head()
```

```python
import pandas as pd
import numpy as np

# Target size
target_size = 1000

# Create new rows by sampling (with variation)
df_expanded = df.sample(n=target_size, replace=True).reset_index(drop=True)

# Optional: Add small randomness to Savings (to avoid exact duplicates)
df_expanded["Savings"] = df_expanded["Savings"] + np.random.randint(-200, 500, size=target_size)
df_expanded["Savings"] = df_expanded["Savings"].clip(lower=0)

# Save new dataset
df_expanded.to_csv("clean_data.csv", index=False)

print("Done! 1000 rows dataset created.")
```
